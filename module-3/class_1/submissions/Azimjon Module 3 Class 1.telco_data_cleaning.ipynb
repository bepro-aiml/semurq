```python
import pandas as pd
import numpy as np
```

## Task 1: Load and Inspect the Dataset

```python
# Upload the file WA_Fn-UseC_-Telco-Customer-Churn.csv to Colab first
df = pd.read_csv('WA_Fn-UseC_-Telco-Customer-Churn.csv')
print("Dataset Shape:", df.shape)
```

```python
print("\nData Types:")
print(df.dtypes)
print("\n")
df.info()
```

```python
print("\nNumerical Summary:")
print(df.describe())
print("\nCategorical Summary:")
print(df.describe(include='object'))
```

## Task 2: Identify and Handle Missing Values

```python
# Check missing values
print(df.isnull().sum())

# Check hidden blank spaces
print("\nBlank strings in TotalCharges:", len(df[df['TotalCharges'] == ' ']))
```

```python
# Fix TotalCharges
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')

# Create missing indicator
df['TotalCharges_missing'] = df['TotalCharges'].isnull().astype(int)

# Impute with median
df['TotalCharges'].fillna(df['TotalCharges'].median(), inplace=True)
```

## Task 3: Standardize Inconsistent Categorical Entries

```python
# Strip whitespace from all object columns
for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].str.strip()
```

## Task 4: Convert Data Types and Extract Temporal Features

```python
# Create temporal feature
df['tenure_years'] = df['tenure'] / 12
```

## Task 5: Detect and Remove Duplicates

```python
print("Number of duplicate rows:", df.duplicated().sum())
# df = df.drop_duplicates()   # Uncomment if duplicates exist
```

## Task 6: Identify Unrealistic Values

```python
print("Negative MonthlyCharges:", len(df[df['MonthlyCharges'] < 0]))
print("Tenure > 72 months:", len(df[df['tenure'] > 72]))
```

## Final Check and Save

```python
print("\nFinal Dataset Shape:", df.shape)
df.info()

# Save the cleaned dataset
df.to_csv('telco_cleaned.csv', index=False)
print("\n✅ Cleaned dataset saved as 'telco_cleaned.csv'")
```

**Justification for Submission**

- **Missing Values**: `TotalCharges` had 11 blank entries (MNAR - related to tenure=0). Added missing indicator and used median imputation.
- Cleaned string columns by stripping whitespace.
- Added `tenure_years` feature.
- No duplicates or unrealistic values found.
- All columns now have appropriate data types.
