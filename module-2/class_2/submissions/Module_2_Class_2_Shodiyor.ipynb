```python
import pandas as pd
df = pd.read_csv('SampleSuperstore.csv', encoding="latin1")
df.head(10)
```

Bu qatorda datasetni tekshirib oldik nimalar bor yoki yo'qligini

```python
print("Shape:", df.shape)
print("Info:", df.info())
print("Describe:", df.describe())
print(df.isnull().sum())
print("Duplicate rows:", df.duplicated().sum())
```

bu qatroda datesetning o'lchamini (shape) aniqladik va asosiy ma'lumotlarni oldik uning string, object, int medianasi, mean yoki min/max, bo'sh qiymatlar, nusxalangan qatorlarni aniqlab oldik

```python
print(df.isnull().sum())
```

```python
df['Sales'] = df['Sales'].fillna(df['Sales'].median())
```

```python
df['Ship Mode'].fillna(df['Ship Mode'].mode()[0], inplace=True)
```

bu 2 qatorda Sales va Ship modedagi bo'sh kataklarni to'ldirik. Salesda bo'sh bo'lsa uning medianasi yozildi, Ship mode'da mode, yani medianada o'rtadagi qiymatni yozgan bo'lsa mode'da eng ko'p takrorlangani yozildi.

```python
print(df.isnull().sum())
```

there was no missing value

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()
print("Shapes after removing duplicates:", df.shape)
```

datasetda duplicatlar yoq va drop duplicate - duplicatlarni olib tashlagandan so'ng datasetning o'lchami(shape) tekshirildi va duplicates = 0 bo'lgani uchun shape ham o'zgarmadi

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])
print(df.dtypes)
df[['Order Date', 'Ship Date']].head(10)
```

order date va ship date larni datetimega o'zgartirdik. birinchi qatordagi kodga qarasak datasetdagi  sanalar  quyidagicha: 1/8/2016. bu qatordagi koddan so'ng sanalar quyidagicha: 2016-11-08. bundan maqsad avval string holatda bo'lsa hozir datetime holatda bu orqali analizda string holatda to'g'ri malumot ola olmaydi shuning uchun uni datetimega o'tkazib to'girlab qo'ydik

```python
total_spending = df.groupby('Customer ID')['Sales'].sum().reset_index()
total_spending.columns = ['Customer ID', 'total_spending']
total_spending.head()
order_freq = df.groupby('Customer ID')['Order ID'].nunique().reset_index()
order_freq.columns = ['Customer ID', 'order_frequency']
order_freq.head(8)
avg_order = df.groupby('Customer ID')['Sales'].mean().reset_index()
avg_order.columns = ['Customer ID', 'avg_order_value']
avg_order.head(8)
customer_summary = total_spending.merge(order_freq, on='Customer ID') \
                                 .merge(avg_order, on='Customer ID')
```

bunda har bir xaridor bir necha marotaba buyurtma bergan bo'lishi mumkin. shuning uchun biz groupby orqali bir xaridor umumiy necha marta buyurtma berganini nunique orqali aniqladik, .sum orqali umumiy qancha summaga buyurtma berganini aniqladik, mean orqali har bir buyurtmasi o'rtacha qanchaga tushganini aniqladik. shunda bizda A B C D ... xaridorlar bolsa, A (B, C...) xaridor necha marta buyurtma berdi, umumiy summa qancha bo'ldi, har bir buyurtma o'rtacha qanchaga tushganini yangi jadvalda aks ettrirldi va yangi fayl qilib saqlandi.

```python
df.to_csv('superstore_cleaned.csv', index=False)
customer_summary.to_csv('customer_summary.csv', index=False)
```
