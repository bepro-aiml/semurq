```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load cleaned dataset
df = pd.read_csv('superstore_cleaned.csv')

# Convert Order Date to datetime
df['Order Date'] = pd.to_datetime(df['Order Date'])

# Display first rows
df.head()
```

## Task 1: Distribution Plots

```
plt.figure(figsize=(10, 6))
plt.hist(df['Sales'], bins=50, color='steelblue', edgecolor='black')

plt.title('Distribution of Sales')
plt.xlabel('Sales ($)')
plt.ylabel('Frequency')

plt.show()
```

### Interpretation

The Sales distribution is right-skewed, meaning most transactions are small while a few orders have very large sales values. This is common in retail datasets because a small number of customers or orders usually contribute a large portion of revenue. The histogram also shows that extremely high sales values are relatively rare.


```
plt.figure(figsize=(10, 6))
sns.boxplot(x=df['Profit'], color='lightcoral')

plt.title('Boxplot of Profit')
plt.xlabel('Profit ($)')

plt.show()
```

### Interpretation

The boxplot shows the distribution of Profit values and highlights the median and quartiles clearly. There are many visible outliers on both the positive and negative sides, indicating some transactions generated unusually high profits while others caused significant losses. The spread also suggests that profit values vary widely across orders.


## Task 2: Category-Wise Trends

```
cat_sales = df.groupby('Category')['Sales'].sum().sort_values(ascending=False)

plt.figure(figsize=(8, 5))
cat_sales.plot(kind='bar', color=['#2196F3', '#4CAF50', '#FF9800'])

plt.title('Total Sales by Category')
plt.xlabel('Category')
plt.ylabel('Total Sales ($)')
plt.xticks(rotation=0)

plt.show()
```

```
region_sales = df.groupby('Region')['Sales'].sum().sort_values(ascending=False)

plt.figure(figsize=(8, 5))
region_sales.plot(kind='bar', color='steelblue')

plt.title('Total Sales by Region')
plt.xlabel('Region')
plt.ylabel('Total Sales ($)')
plt.xticks(rotation=0)

plt.show()
```

### Interpretation

The Technology category usually generates the highest sales revenue because technology products tend to have higher prices compared to office supplies and furniture. Among regions, the West region often leads in sales performance. The charts make it easy to compare categories and regions visually and identify the strongest contributors to total revenue.


## Task 3: Outlier Identification

```
Q1 = df['Profit'].quantile(0.25)
Q3 = df['Profit'].quantile(0.75)
IQR = Q3 - Q1

print(f"Q1: {Q1:.2f}")
print(f"Q3: {Q3:.2f}")
print(f"IQR: {IQR:.2f}")

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(f"Lower bound: {lower_bound:.2f}")
print(f"Upper bound: {upper_bound:.2f}")

outliers = df[(df['Profit'] < lower_bound) | (df['Profit'] > upper_bound)]

print(f"Number of outliers: {len(outliers)} out of {len(df)} rows ({len(outliers)/len(df)*100:.1f}%)")
```

### Interpretation

The IQR method identifies transactions with unusually high or low profits as outliers. Most outliers in retail data are often associated with large discounts, bulk purchases, or expensive products. Some outliers represent very high profits, while others indicate major losses from heavily discounted orders.


## Task 4: Correlation Heatmap

```
numerical_cols = df[['Sales', 'Quantity', 'Discount', 'Profit']]

corr_matrix = numerical_cols.corr()

print(corr_matrix)

plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', center=0, fmt='.2f')

plt.title('Correlation Heatmap: Sales, Quantity, Discount, Profit')

plt.show()
```

### Interpretation

The heatmap shows relationships between numerical variables in the dataset. Sales and Profit usually have a positive correlation, meaning larger sales often lead to higher profits. Discount and Profit commonly show a negative correlation, suggesting that larger discounts may reduce profitability.


## Task 5: Time-Based Trend

```
monthly_sales = df.resample('M', on='Order Date')['Sales'].sum()

plt.figure(figsize=(12, 6))
monthly_sales.plot(color='steelblue', linewidth=2)

plt.title('Monthly Total Sales Over Time')
plt.xlabel('Month')
plt.ylabel('Total Sales ($)')
plt.grid(True, alpha=0.3)

plt.show()
```

### Interpretation

The monthly sales trend helps identify seasonal patterns and changes over time. Retail businesses often experience higher sales near the end of the year because of holiday shopping and promotions. Any visible dips may indicate slower business periods or seasonal demand changes.


## Task 6: Reflection Questions

### 1. Why is the Sales histogram right-skewed in retail data?

Retail datasets usually contain many small purchases and a few extremely large orders. Because of this, the majority of sales values stay near the lower end while a small number of transactions create a long tail on the right side of the distribution.

### 2. Should outliers in Profit be removed?

Not always. Outliers may contain important business information such as unusually profitable orders or major losses caused by discounts or returns. Removing them could hide valuable patterns and reduce the accuracy of business analysis.

### 3. Does a negative Discount-Profit correlation mean discounts should stop?

No. Discounts can still increase customer demand, attract new buyers, and improve long-term sales. The correlation only suggests that larger discounts are often associated with lower profits, not that discounts are always harmful.
