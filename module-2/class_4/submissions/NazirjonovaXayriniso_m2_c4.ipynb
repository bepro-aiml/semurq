## Step 0 — Load the Data

```
# Import pandas library for data analysis
import pandas as pd

# Load the Telco Customer Churn dataset from GitHub
url = 'https://raw.githubusercontent.com/IBM/telco-customer-churn-on-icp4d/master/data/Telco-Customer-Churn.csv'
df = pd.read_csv(url)

# Convert TotalCharges column to numeric values and replace missing values with 0
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce').fillna(0)

# Print the shape of the dataset
print('Shape:', df.shape)

# Display first rows of the dataset
df.head()
```

## Step 1 — See Column Information

```
# Display information about columns, data types, and missing values
df.info()
```

## Step 2 — Statistics for Numeric Columns

```
# Display summary statistics for numerical columns
df.describe().round(2)
```

## Step 3 — Count Unique Values per Column

```
# Count the number of unique values in each column
df.nunique()
```

## Step 4 — Filter Customers Paying More Than $80/Month

```
# Filter customers with MonthlyCharges greater than 80
high_pay = df[df['MonthlyCharges'] > 80]

# Print the number of high-paying customers
print('High-paying customers:', len(high_pay))

# Display the first rows of filtered data
high_pay.head()
```

### Explanation

This filtering operation selects customers who pay more than $80 per month.
Filtering is useful for identifying specific customer groups and analyzing them separately.


## Step 5 — Sort Top 10 Customers by Tenure

```
# Sort customers by tenure in descending order and select top 10
top10 = df.sort_values('tenure', ascending=False).head(10)

# Display selected columns for the top 10 customers
top10[['customerID', 'tenure', 'Contract', 'MonthlyCharges']]
```

### Explanation

Sorting helps organize the dataset based on specific columns.
Here, customers with the highest tenure are shown first to identify long-term customers.


## Step 6 — Average Monthly Charges per Payment Method

```
# Group data by PaymentMethod and calculate average MonthlyCharges
df.groupby('PaymentMethod')['MonthlyCharges'].mean().round(2)
```

### Explanation

Grouping allows comparison between categories.
This analysis shows how average monthly charges vary across payment methods.


## Step 7 — Groupby with Two Statistics

```
# Group by Churn and calculate mean and count of TotalCharges
df.groupby('Churn').agg({'TotalCharges': ['mean', 'count']}).round(2)
```

### Explanation

Aggregation functions summarize important information from the dataset.
This comparison helps analyze differences between customers who churned and those who stayed.


## Step 8 — Crosstab: Contract vs Churn

```
# Create a crosstab comparing Contract type and Churn
pd.crosstab(df['Contract'], df['Churn'])
```

### Explanation

A crosstab compares relationships between categorical variables.
This table shows how churn behavior changes depending on contract type.
