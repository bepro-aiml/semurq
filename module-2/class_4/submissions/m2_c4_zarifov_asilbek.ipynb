## **Section3** — Filtering, GroupBy, Merging (maps to lecture section “Filtering,GroupBy, Merging”

### **Task 3.1 — Selection**


```python
# Second data
#Load any small CSV. If you don't have one, use this built-in dataset:
from sklearn.datasets import load_iris
iris = load_iris(as_frame=True)
df1 = iris.frame
df1.head()
```

```python
df1["sepal length (cm)"]
```

```python
df1[["sepal length (cm)", "petal length (cm)"]]
```

```python
df1.loc[0].head(10)
```

```python
df1.iloc[0]
```

### **Task 3.2 — Boolean filtering**

```python
df1[(df1["sepal length (cm)"] > 5) & (df1["target"] == 0)]
```

```python
df1[(df1["petal length (cm)"] < 2) | (df1["sepal width (cm)"] > 3.5)]
```

### **Task 3.3 — GroupBy: split-apply-combine**

```python
df1.groupby("target")["sepal length (cm)"].mean()
# Simple GroupBy pattern (mean)
```

```python
df1.groupby("target").agg({
    "sepal length (cm)": "mean",
    "petal length (cm)": "count"
})

# GroupBy with .agg() (multiple calculations)
```

### **Task 3.4 — Merging**

```python
lookup_df = pd.DataFrame({
    "target": [0, 1, 2],
    "species_name": ["setosa", "versicolor", "virginica"]
})

lookup_df
# We will use the Iris dataset (df1) and create a lookup table for target → species name
```

```python
merged_df = pd.merge(df1, lookup_df, on="target", how="left")

merged_df.head()
# Step 2: Merge with main DataFrame
```
